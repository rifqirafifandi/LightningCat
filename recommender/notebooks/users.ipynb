```python
from faker import Faker
import random
from faker.providers import misc, DynamicProvider, geo
import json

activity_map = {
    "SWIMMING_C": "Swimming",
    "FOOTBALL_S": "Football",
    "TENNIS_SQU": "Tennis",
    "GYM_OPERAT": "Gym",
    "WADING_POO": "Wading",
    "INDOOR_SPO": "Indoor",
    "BADMINTON_": "Badminton",
    "TABLE_TENN": "Table_tennis",
    "NETBALL_CO": "Netball",
    "VOLLEYBALL": "Volleyball",
    "BASKETBALL": "Basketball",
    "ATHLETICS_": "Athletics",
    "FOOTBALL_F": "Football",
    "SOCCER_COU": "Soccer",
    "RUGBY_FIEL": "Rugby",
    "TENNIS_COU": "Tennis",
    "SQUASH_COU": "Squash",
    "GYM": "Gym",
    "HOCKEY_PIT": "Hockey",
    "PETANQUE_C": "Petanque",
    "GATEBALL_C": "Gateball",
    "LAWN_BOWL_": "Lawn_bowl",
    "PICKLEBALL": "Pickleball",
}

ACTIVITIES = list(activity_map.values())
COORDS = [
    [1.30962889460807, 103.762020397147],
    [1.31087340216251, 103.765697311082],
    [1.30981721657829, 103.766484018075],
    [1.33893979108182, 103.693702605961],
    [1.33971970562101, 103.694861154421],
    [1.33799861493825, 103.693284554819],
]

fake = Faker()
fake.add_provider(misc)
fake.add_provider(geo)

age_range_provider = DynamicProvider(
    provider_name="age_range",
    elements=["Children", "Adult", "Senior"],
)

activity_provider = DynamicProvider(
    provider_name="activity",
    elements=ACTIVITIES,
)

latlng_provider = DynamicProvider(
    provider_name="sg_latlng",
    elements=COORDS,
)

fake.add_provider(age_range_provider)
fake.add_provider(activity_provider)
fake.add_provider(latlng_provider)
```

```python
users = []

for _ in range(10):
    username = fake.uuid4()
    activities = [fake.unique.activity() for _ in range(random.randint(1, 4))]
    age_range = fake.age_range()
    location = fake.sg_latlng()
    user = {
        "username": username,
        "activities": activities,
        "age_range": age_range,
        "location": location,
    }
    fake.unique.clear()
    users.append(user)
```

```python
json.dump(users, open("user_data.json", "w"), indent=4)
```
