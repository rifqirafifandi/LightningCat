```python
import json
from shapely.geometry import Point
from shapely.geometry.polygon import Polygon
```

```python
plan_area = json.load(
    open(
        "/Users/siwei/dev/LightningCat/dataAndDataCleanup/MasterPlan2019PlanningAreaBoundaryNoSea.processed.geojson"
    )
)
```

```python
facilities = json.load(
    open("/Users/siwei/dev/LightningCat/dataAndDataCleanup/facility_data.json")
)
```

```python
plan_polygon = {}
for plan in plan_area["features"]:
    name = plan["properties"]["PLN_AREA_N"]
    ps = []
    coordinates = plan["geometry"]["coordinates"][0]
    if not isinstance(coordinates[0][0], float):
        coordinates = coordinates[0]
    for c in coordinates:
        ps.append((c[1], c[0]))
    plan_polygon[name.lower()] = Polygon(ps)
```

```python
facility_map = {}
for fa in facilities:
    p = Point(fa["location"])
    for plan_name, polygon in plan_polygon.items():
        if polygon.contains(p):
            facility_map[fa["name"]] = plan_name
            break
```
